## Experiment 4 – Iris Dataset Preprocessing

```python
import pandas as pd
from ucimlrepo import fetch_ucirepo

iris = fetch_ucirepo(id=53)
X = iris.data.features
y = iris.data.targets

print("Feature dimensions:", X.shape)
print("Target dimensions:", y.shape)

display(X.head())
display(y.head())
```

```python
print("Missing values in features:")
print(X.isnull().sum())
print("Missing values in target:")
print(y.isnull().sum())
```

```python
data = pd.concat([X, y], axis=1)
data = data.drop_duplicates()
X = data.iloc[:, :-1]
y = data.iloc[:, -1:]

X = X.fillna(X.mean(numeric_only=True))
y = y.fillna(y.mode().iloc[0])

print("Missing values after cleaning:", X.isnull().sum().sum() + y.isnull().sum().sum())
print("Cleaned feature dimensions:", X.shape)
print("Cleaned target dimensions:", y.shape)

display(pd.concat([X, y], axis=1).head())
```
