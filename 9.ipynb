## Experiment 9 – Support Vector Machine (SVM)

```python
from sklearn import datasets
from sklearn import svm
import matplotlib.pyplot as plt

iris = datasets.load_iris()
X = iris.data
y = iris.target

print(iris.DESCR[:300])
print(X.shape)
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.xlabel("Sepal length")
plt.ylabel("Sepal width")
plt.title("Sepal Features")
plt.show()
```

```python
plt.scatter(X[:, 2], X[:, 3], c=y)
plt.xlabel("Petal length")
plt.ylabel("Petal width")
plt.title("Petal Features")
plt.show()
```

```python
X_sepal = X[:, :2]

linear = svm.SVC(kernel="linear")
rbf = svm.SVC(kernel="rbf")
poly = svm.SVC(kernel="poly", degree=3)

linear.fit(X_sepal, y)
rbf.fit(X_sepal, y)
poly.fit(X_sepal, y)

print("SVM models trained using sepal features")
```

```python
import numpy as np

models = [linear, rbf, poly]
names = ["Linear Kernel", "RBF Kernel", "Polynomial Kernel"]

x_min = X_sepal[:, 0].min() - 1
x_max = X_sepal[:, 0].max() + 1
y_min = X_sepal[:, 1].min() - 1
y_max = X_sepal[:, 1].max() + 1

xx, yy = np.meshgrid(
    np.arange(x_min, x_max, 0.05),
    np.arange(y_min, y_max, 0.05)
)

for model, name in zip(models, names):
    Z = model.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X_sepal[:, 0], X_sepal[:, 1], c=y)
    plt.xlabel("Sepal length")
    plt.ylabel("Sepal width")
    plt.title(name)
    plt.show()
```
