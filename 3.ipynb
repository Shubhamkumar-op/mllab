## Experiment 3 – Data Visualization

This experiment reads `student_performance_dataset.csv`, as shown in the PDF.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student_performance_dataset.csv")
df.head()
```

```python
df.columns
```

```python
plt.plot(df.iloc[0:20, 0], df.select_dtypes(include="number").iloc[0:20, 0])
plt.title("Line Graph")
plt.xlabel("Student Index")
plt.ylabel("Value")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
columns = numeric_df.columns
for column in columns[:3]:
    plt.plot(numeric_df.index[:20], numeric_df[column].iloc[:20], label=column)
plt.title("Multi-Line Graph")
plt.xlabel("Student Index")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
category = df.select_dtypes(include="object").columns[0]
df[category].value_counts().head(6).plot(kind="pie", autopct="%1.1f%%")
plt.title("Pie Chart")
plt.ylabel("")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
sns.scatterplot(data=df, x=numeric_df.columns[0], y=numeric_df.columns[1])
plt.title("Scatter Plot")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
sns.boxplot(data=numeric_df)
plt.title("Box Plot")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
numeric_df.iloc[:, :5].plot(kind="bar")
plt.title("Bar Graph")
plt.xlabel("Student Index")
plt.ylabel("Value")
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
for column in numeric_df.columns[:5]:
    plt.hist(numeric_df[column].dropna(), bins=10, alpha=0.5, label=column)
plt.title("Histogram")
plt.legend()
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Heatmap")
plt.show()
```
